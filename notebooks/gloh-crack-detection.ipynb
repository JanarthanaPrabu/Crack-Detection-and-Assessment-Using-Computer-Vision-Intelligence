```python
import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import io
from PIL import Image
import ipywidgets as widgets
from IPython.display import display, clear_output

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.pipeline import make_pipeline
from sklearn.cluster import MiniBatchKMeans
from sklearn.metrics import (
    accuracy_score, precision_recall_fscore_support, confusion_matrix
)
from sklearn.utils.class_weight import compute_sample_weight

from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier

# =====================================================================
# CONFIGURATION SETTINGS
# =====================================================================
TEST_SIZE = 0.30          # 70/30 Train/Test split
SAMPLES_PER_CLASS = 2000  # Cap per class for balanced training & fast run
VOCAB_SIZE = 128          # Bag-of-Visual-Words codebook dimension (128-D output)
PATCH_DIM = 32            # Neighborhood patch size around keypoints

# OpenCV SIFT Detector for Keypoints (compatible with modern OpenCV versions)
sift_detector = cv2.SIFT_create(nfeatures=150, contrastThreshold=0.01, edgeThreshold=10)

# =====================================================================
# 1. FIXED LOG-POLAR GLOH FEATURE EXTRACTION ENGINE
# =====================================================================
def create_gloh_spatial_grid(patch_size=32):
    """
    Creates GLOH's 17 Spatial Log-Polar Regions:
      - Region 0: Central Circle
      - Regions 1..8: Inner Ring (8 angular sectors)
      - Regions 9..16: Outer Ring (8 angular sectors)
    """
    h, w = patch_size, patch_size
    cy, cx = h // 2, w // 2
    
    y, x = np.ogrid[:h, :w]
    radii = np.sqrt((x - cx)**2 + (y - cy)**2)
    angles_deg = np.degrees(np.arctan2(y - cy, x - cx)) % 360

    max_radius = min(cx, cy)
    r0 = max_radius * 0.25
    r1 = max_radius * 0.65
    r2 = max_radius

    grid_map = np.full((h, w), -1, dtype=np.int32)
    grid_map[radii <= r0] = 0

    sector = np.floor(angles_deg / 45.0).astype(np.int32) % 8

    inner_mask = (radii > r0) & (radii <= r1)
    grid_map[inner_mask] = 1 + sector[inner_mask]

    outer_mask = (radii > r1) & (radii <= r2)
    grid_map[outer_mask] = 9 + sector[outer_mask]

    return grid_map

GLOH_GRID = create_gloh_spatial_grid(PATCH_DIM)


def extract_gloh_descriptors_from_image(img_path, patch_dim=32):
    """
    Fixed Implementation incorporating your guide's workflow:
    1. Read Image
    2. Sobel Gradients -> Polar Magnitude & Orientation Angle
    3. SIFT Keypoint Detection
    4. Log-Polar GLOH Histogram Calculation per Keypoint
    """
    image = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    if image is None:
        return None

    # Compute gradient magnitude and orientation using Sobel operators
    grad_x = cv2.Sobel(image, cv2.CV_32F, 1, 0, ksize=3)
    grad_y = cv2.Sobel(image, cv2.CV_32F, 0, 1, ksize=3)
    mag, angle = cv2.cartToPolar(grad_x, grad_y, angleInDegrees=True)

    # Bin angles into 8 orientation channels (0..7)
    angle_bins = (np.floor(angle / 45.0).astype(np.int32)) % 8

    # Detect SIFT keypoints
    keypoints = sift_detector.detect(image, None)
    if not keypoints:
        return np.zeros((1, 128), dtype=np.float32)

    half_p = patch_dim // 2
    h_img, w_img = image.shape
    gloh_descriptors = []

    # Compute GLOH features for each keypoint with safe boundary checking
    for kp in keypoints:
        x, y = int(kp.pt[0]), int(kp.pt[1])
        y_min, y_max = y - half_p, y + half_p
        x_min, x_max = x - half_p, x + half_p

        # Boundary check to prevent slicing errors
        if y_min < 0 or y_max > h_img or x_min < 0 or x_max > w_img:
            continue

        mag_patch = mag[y_min:y_max, x_min:x_max]
        bin_patch = angle_bins[y_min:y_max, x_min:x_max]

        # 17 spatial regions x 8 orientation bins
        hist_matrix = np.zeros((17, 8), dtype=np.float32)
        valid_px = GLOH_GRID >= 0
        
        # Accumulate gradient magnitude into spatial bins
        np.add.at(
            hist_matrix, 
            (GLOH_GRID[valid_px], bin_patch[valid_px]), 
            mag_patch[valid_px]
        )

        # Flatten into descriptor vector & truncate to 128 dimensions (GLOH standard)
        desc = hist_matrix.ravel()[:128]
        
        # L2 Normalize individual keypoint descriptor
        norm = np.linalg.norm(desc)
        if norm > 0:
            desc = desc / norm

        gloh_descriptors.append(desc)

    if not gloh_descriptors:
        return np.zeros((1, 128), dtype=np.float32)

    return np.array(gloh_descriptors, dtype=np.float32)

# =====================================================================
# 2. EXACT MATCH PARSER FOR SDNET DATASET
# =====================================================================
def parse_sdnet_labels(path):
    p = path.replace('\\', '/').lower()
    parts = p.split('/')
    
    leaf_folder = parts[-2] if len(parts) >= 2 else ""
    parent_folder = parts[-3] if len(parts) >= 3 else ""
    
    struct = None
    status = None

    if "non-cracked" in leaf_folder or "uncracked" in leaf_folder:
        status = "No Crack"
    elif "cracked" in leaf_folder:
        status = "Crack"

    if "deck" in parent_folder or "deck" in leaf_folder:
        struct = "Bridge Deck"
    elif "pavement" in parent_folder or "pavement" in leaf_folder:
        struct = "Pavement"
    elif "wall" in parent_folder or "wall" in leaf_folder:
        struct = "Wall"

    if struct is None or status is None:
        return None

    return status, struct, f"{struct} | {status}"

# Discover images
print("🔍 Discovering images in Kaggle directory...")
all_jpgs = []
for root, dirs, files in os.walk('/kaggle/input'):
    for f in files:
        if f.lower().endswith(('.jpg', '.jpeg', '.png')):
            all_jpgs.append(os.path.join(root, f))

# =====================================================================
# 3. VISUALIZE REPRESENTATIVE SAMPLE IMAGE FOR EVERY DETECTED CLASS
# =====================================================================
print("\n🖼️ Displaying sample labeled images from dataset folders...")
class_samples = {}
for img_path in all_jpgs:
    parsed = parse_sdnet_labels(img_path)
    if parsed:
        l6 = parsed[2]
        if l6 not in class_samples:
            class_samples[l6] = img_path

fig, axes = plt.subplots(2, 3, figsize=(12, 8))
axes = axes.ravel()

for idx, (label_name, sample_path) in enumerate(class_samples.items()):
    if idx < 6:
        sample_img = cv2.imread(sample_path)
        sample_img = cv2.cvtColor(sample_img, cv2.COLOR_BGR2RGB)
        axes[idx].imshow(sample_img)
        axes[idx].set_title(f"Label: {label_name}", fontsize=11, fontweight='bold', color='navy')
        axes[idx].axis('off')

plt.tight_layout()
plt.show()

# =====================================================================
# 4. FEATURE EXTRACTION & BOVW QUANTIZATION PIPELINE
# =====================================================================
raw_descriptors_list, image_paths = [], []
y_2class_raw, y_3class_raw, y_6class_raw = [], [], []
class_counts = {}

np.random.seed(42)
np.random.shuffle(all_jpgs)

print(f"\n⚙️ Extracting GLOH descriptors for dataset images...")
sampled_descriptors_for_kmeans = []

for img_path in tqdm(all_jpgs):
    parsed = parse_sdnet_labels(img_path)
    if parsed is None:
        continue
    
    l2, l3, l6 = parsed
    if class_counts.get(l6, 0) >= SAMPLES_PER_CLASS:
        continue

    desc = extract_gloh_descriptors_from_image(img_path)
    if desc is not None:
        raw_descriptors_list.append(desc)
        y_2class_raw.append(l2)
        y_3class_raw.append(l3)
        y_6class_raw.append(l6)
        image_paths.append(img_path)
        class_counts[l6] = class_counts.get(l6, 0) + 1

        # Subsample descriptors to build visual codebook efficiently
        if len(desc) > 20:
            idx = np.random.choice(len(desc), 20, replace=False)
            sampled_descriptors_for_kmeans.append(desc[idx])
        else:
            sampled_descriptors_for_kmeans.append(desc)

print(f"\n📊 Training Codebook (MiniBatchKMeans Vocabulary Size = {VOCAB_SIZE})...")
kmeans_data = np.vstack(sampled_descriptors_for_kmeans)
kmeans = MiniBatchKMeans(n_clusters=VOCAB_SIZE, random_state=42, batch_size=4096, n_init="auto")
kmeans.fit(kmeans_data)
del kmeans_data, sampled_descriptors_for_kmeans

print("\n📐 Quantizing GLOH descriptors into BoVW Histograms (128-D Fixed Vector per Image)...")
X = np.zeros((len(raw_descriptors_list), VOCAB_SIZE), dtype=np.float32)
for i, desc in enumerate(tqdm(raw_descriptors_list)):
    clusters = kmeans.predict(desc)
    hist, _ = np.histogram(clusters, bins=np.arange(VOCAB_SIZE + 1))
    norm = np.linalg.norm(hist)
    X[i] = hist / norm if norm > 0 else hist

y_2class = np.array(y_2class_raw)
y_3class = np.array(y_3class_raw)
y_6class = np.array(y_6class_raw)

print("\n--- Extracted Dataset Breakdown (6-Class) ---")
for k, v in class_counts.items():
    print(f" • {k}: {v} images")

# =====================================================================
# 5. PIPELINE TRAINER WITH PER-CLASS BREAKDOWN & MACRO METRICS
# =====================================================================
def train_and_eval_task(task_name, X_data, y_data, test_ratio=0.30):
    train_pct = int((1 - test_ratio) * 100)
    test_pct = int(test_ratio * 100)
    
    print("\n" + "="*70)
    print(f" 🚀 PIPELINE FOR {task_name.upper()} CLASSIFICATION (Split: {train_pct}/{test_pct})")
    print("="*70)

    le = LabelEncoder()
    y_encoded = le.fit_transform(y_data)

    X_train, X_test, y_train, y_test = train_test_split(
        X_data, y_encoded, test_size=test_ratio, random_state=42, stratify=y_encoded
    )

    num_classes = len(np.unique(y_encoded))
    pos_weight = (y_train == 0).sum() / (y_train == 1).sum() if num_classes == 2 else 1.0

    xgb_clf = XGBClassifier(
        n_estimators=200, 
        learning_rate=0.08, 
        eval_metric='mlogloss', 
        scale_pos_weight=pos_weight if num_classes == 2 else 1.0,
        random_state=42, 
        tree_method='hist',
        device='cuda'
    )

    models = {
        "SVM": make_pipeline(StandardScaler(), SVC(kernel='rbf', C=5.0, class_weight='balanced', probability=True, random_state=42)),
        "Random Forest": make_pipeline(StandardScaler(), RandomForestClassifier(n_estimators=200, max_depth=20, class_weight='balanced', random_state=42, n_jobs=-1)),
        "XGBoost": make_pipeline(StandardScaler(), xgb_clf),
        "Decision Tree": make_pipeline(StandardScaler(), DecisionTreeClassifier(max_depth=12, class_weight='balanced', random_state=42))
    }

    trained = {}
    metrics = {}

    for name, model in models.items():
        try:
            if name == "XGBoost" and num_classes > 2:
                sample_weights = compute_sample_weight('balanced', y_train)
                model.fit(X_train, y_train, xgbclassifier__sample_weight=sample_weights)
            else:
                model.fit(X_train, y_train)
        except Exception:
            if name == "XGBoost":
                model.named_steps['xgbclassifier'].set_params(device='cpu')
                if num_classes > 2:
                    sample_weights = compute_sample_weight('balanced', y_train)
                    model.fit(X_train, y_train, xgbclassifier__sample_weight=sample_weights)
                else:
                    model.fit(X_train, y_train)

        preds = model.predict(X_test)
        acc = accuracy_score(y_test, preds)
        
        macro_p, macro_r, macro_f1, _ = precision_recall_fscore_support(
            y_test, preds, average='macro', zero_division=0
        )
        macro_sens = macro_r

        per_p, per_r, per_f1, _ = precision_recall_fscore_support(
            y_test, preds, average=None, zero_division=0
        )
        
        cm = confusion_matrix(y_test, preds)

        trained[name] = model
        metrics[name] = {
            "accuracy": acc, "precision": macro_p, "recall": macro_r, 
            "sensitivity": macro_sens, "f1": macro_f1, "cm": cm
        }

        print(f"\n---> {name} [{task_name}]")
        print(f" • Overall Accuracy   : {acc * 100:.2f}%")
        print(f" • Macro Precision    : {macro_p:.4f}")
        print(f" • Macro Sensitivity  : {macro_sens:.4f}  (Sensitivity = Recall)")
        print(f" • Macro F1-Score     : {macro_f1:.4f}\n")
        print("    --- Per-Class Performance Breakdown ---")
        for cls_idx, cls_name in enumerate(le.classes_):
            p_val = per_p[cls_idx]
            r_val = per_r[cls_idx]
            f_val = per_f1[cls_idx]
            print(f"   - {cls_name:<20} | Precision: {p_val:.4f} | Sensitivity/Recall: {r_val:.4f} | F1: {f_val:.4f}")

    fig, axes = plt.subplots(2, 2, figsize=(10, 8))
    axes = axes.ravel()
    for i, (name, m_info) in enumerate(metrics.items()):
        sns.heatmap(m_info['cm'], annot=True, fmt='d', cmap='Blues',
                    xticklabels=le.classes_, yticklabels=le.classes_, ax=axes[i])
        axes[i].set_title(f"{name} ({task_name})", fontweight='bold')
        axes[i].set_xlabel("Predicted")
        axes[i].set_ylabel("True")
        axes[i].tick_params(axis='x', rotation=45)

    plt.tight_layout()
    plt.show()

    return trained, metrics, le

# Run training pipelines
models_2c, metrics_2c, le_2c = train_and_eval_task("2-Class (Condition)", X, y_2class, test_ratio=TEST_SIZE)
models_3c, metrics_3c, le_3c = train_and_eval_task("3-Class (Surface Structure)", X, y_3class, test_ratio=TEST_SIZE)
models_6c, metrics_6c, le_6c = train_and_eval_task("6-Class (Fine-Grained)", X, y_6class, test_ratio=TEST_SIZE)

# =====================================================================
# 6. KAGGLE UPLOADER WIDGET
# =====================================================================
def extract_single_image_bovw(img_path):
    desc = extract_gloh_descriptors_from_image(img_path)
    if desc is None:
        return None
    clusters = kmeans.predict(desc)
    hist, _ = np.histogram(clusters, bins=np.arange(VOCAB_SIZE + 1))
    norm = np.linalg.norm(hist)
    return (hist / norm if norm > 0 else hist).reshape(1, -1)


def create_kaggle_uploader(output_folder="/kaggle/working/predictions"):
    os.makedirs(output_folder, exist_ok=True)
    
    upload_widget = widgets.FileUpload(
        accept='.jpg,.jpeg,.png',
        multiple=False,
        description='Upload Image'
    )
    output_area = widgets.Output()

    def on_file_upload(change):
        with output_area:
            clear_output()
            if not upload_widget.value:
                return
            
            uploaded_file = list(upload_widget.value.values())[0] if isinstance(upload_widget.value, dict) else upload_widget.value[0]
            fname = uploaded_file['name'] if isinstance(uploaded_file, dict) else uploaded_file.name
            content = uploaded_file['content'] if isinstance(uploaded_file, dict) else uploaded_file.content
            
            saved_img_path = os.path.join(output_folder, fname)
            with open(saved_img_path, "wb") as f:
                f.write(content)
            
            feats = extract_single_image_bovw(saved_img_path)
            if feats is None:
                print(f"❌ Error: Unable to extract features from uploaded image '{fname}'.")
                return

            print(f"\n=================================================================")
            print(f" 📥 ANALYSIS & STRUCTURE DETECTION FOR: {fname}")
            print(f"=================================================================\n")

            img_pil = Image.open(io.BytesIO(content))
            plt.figure(figsize=(4, 4))
            plt.imshow(img_pil)
            plt.title(f"Target Image: {fname}", fontsize=10, fontweight='bold')
            plt.axis('off')
            plt.show()

            primary_model = models_6c["XGBoost"]
            pred_code = primary_model.predict(feats)[0]
            pred_label = le_6c.inverse_transform([pred_code])[0]
            
            detected_surface, detected_condition = pred_label.split(" | ")

            print("🔍 STRUCTURE DETECTION RESULT:")
            print("-----------------------------------------------------------------")
            print(f" 🏛️ Detected Structure Surface : {detected_surface.upper()} DETECTED")
            print(f" ⚠️ Detected Structural Condition : {detected_condition.upper()}")
            print("-----------------------------------------------------------------\n")

            print("📊 MULTI-MODEL METRIC PERFORMANCE BREAKDOWN:")
            for m_name in models_6c.keys():
                p_2c = le_2c.inverse_transform([models_2c[m_name].predict(feats)[0]])[0]
                p_3c = le_3c.inverse_transform([models_3c[m_name].predict(feats)[0]])[0]
                p_6c = le_6c.inverse_transform([models_6c[m_name].predict(feats)[0]])[0]
                
                print(f"\n🔹 ALGORITHM: {m_name}")
                print(f"   • 2-Class Condition Output : {p_2c}")
                print(f"   • 3-Class Structure Output : {p_3c}")
                print(f"   • 6-Class Full Prediction  : {p_6c}")

            txt_save_path = os.path.join(output_folder, os.path.splitext(fname)[0] + "_report.txt")
            with open(txt_save_path, "w") as f:
                f.write(f"DETECTION REPORT FOR: {fname}\n")
                f.write(f"Structure Detected: {detected_surface}\n")
                f.write(f"Condition State: {detected_condition}\n")

            print(f"\n✅ Classified image report saved at: '{output_folder}'")
            print("=================================================================\n")

    upload_widget.observe(on_file_upload, names='value')
    
    print("\n=================================================================")
    print(" 📥 KAGGLE IMAGE UPLOADER & DETECTOR WIDGET")
    print("=================================================================")
    display(upload_widget)
    display(output_area)

create_kaggle_uploader()
```

```python
import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import io
from PIL import Image
import ipywidgets as widgets
from IPython.display import display, clear_output

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.pipeline import make_pipeline
from sklearn.cluster import MiniBatchKMeans
from sklearn.metrics import (
    accuracy_score, precision_recall_fscore_support, confusion_matrix, balanced_accuracy_score
)
from sklearn.utils.class_weight import compute_sample_weight

from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier

# =====================================================================
# CONFIGURATION SETTINGS
# =====================================================================
TEST_SIZE = 0.30          # 70/30 Train/Test split
SAMPLES_PER_CLASS = 2000  # Cap per class for balanced training & fast run
VOCAB_SIZE = 128          # Bag-of-Visual-Words codebook dimension (128-D output)
PATCH_DIM = 32            # Neighborhood patch size around keypoints

# OpenCV SIFT Detector for Keypoints (compatible with modern OpenCV versions)
sift_detector = cv2.SIFT_create(nfeatures=150, contrastThreshold=0.01, edgeThreshold=10)

# =====================================================================
# 1. FIXED LOG-POLAR GLOH FEATURE EXTRACTION ENGINE
# =====================================================================
def create_gloh_spatial_grid(patch_size=32):
    """
    Creates GLOH's 17 Spatial Log-Polar Regions:
      - Region 0: Central Circle
      - Regions 1..8: Inner Ring (8 angular sectors)
      - Regions 9..16: Outer Ring (8 angular sectors)
    """
    h, w = patch_size, patch_size
    cy, cx = h // 2, w // 2
    
    y, x = np.ogrid[:h, :w]
    radii = np.sqrt((x - cx)**2 + (y - cy)**2)
    angles_deg = np.degrees(np.arctan2(y - cy, x - cx)) % 360

    max_radius = min(cx, cy)
    r0 = max_radius * 0.25
    r1 = max_radius * 0.65
    r2 = max_radius

    grid_map = np.full((h, w), -1, dtype=np.int32)
    grid_map[radii <= r0] = 0

    sector = np.floor(angles_deg / 45.0).astype(np.int32) % 8

    inner_mask = (radii > r0) & (radii <= r1)
    grid_map[inner_mask] = 1 + sector[inner_mask]

    outer_mask = (radii > r1) & (radii <= r2)
    grid_map[outer_mask] = 9 + sector[outer_mask]

    return grid_map

GLOH_GRID = create_gloh_spatial_grid(PATCH_DIM)


def extract_gloh_descriptors_from_image(img_path, patch_dim=32):
    """
    Fixed Implementation incorporating your guide's workflow:
    1. Read Image
    2. Sobel Gradients -> Polar Magnitude & Orientation Angle
    3. SIFT Keypoint Detection
    4. Log-Polar GLOH Histogram Calculation per Keypoint
    """
    image = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    if image is None:
        return None

    # Compute gradient magnitude and orientation using Sobel operators
    grad_x = cv2.Sobel(image, cv2.CV_32F, 1, 0, ksize=3)
    grad_y = cv2.Sobel(image, cv2.CV_32F, 0, 1, ksize=3)
    mag, angle = cv2.cartToPolar(grad_x, grad_y, angleInDegrees=True)

    # Bin angles into 8 orientation channels (0..7)
    angle_bins = (np.floor(angle / 45.0).astype(np.int32)) % 8

    # Detect SIFT keypoints
    keypoints = sift_detector.detect(image, None)
    if not keypoints:
        return np.zeros((1, 128), dtype=np.float32)

    half_p = patch_dim // 2
    h_img, w_img = image.shape
    gloh_descriptors = []

    # Compute GLOH features for each keypoint with safe boundary checking
    for kp in keypoints:
        x, y = int(kp.pt[0]), int(kp.pt[1])
        y_min, y_max = y - half_p, y + half_p
        x_min, x_max = x - half_p, x + half_p

        # Boundary check to prevent slicing errors
        if y_min < 0 or y_max > h_img or x_min < 0 or x_max > w_img:
            continue

        mag_patch = mag[y_min:y_max, x_min:x_max]
        bin_patch = angle_bins[y_min:y_max, x_min:x_max]

        # 17 spatial regions x 8 orientation bins
        hist_matrix = np.zeros((17, 8), dtype=np.float32)
        valid_px = GLOH_GRID >= 0
        
        # Accumulate gradient magnitude into spatial bins
        np.add.at(
            hist_matrix, 
            (GLOH_GRID[valid_px], bin_patch[valid_px]), 
            mag_patch[valid_px]
        )

        # Flatten into descriptor vector & truncate to 128 dimensions (GLOH standard)
        desc = hist_matrix.ravel()[:128]
        
        # L2 Normalize individual keypoint descriptor
        norm = np.linalg.norm(desc)
        if norm > 0:
            desc = desc / norm

        gloh_descriptors.append(desc)

    if not gloh_descriptors:
        return np.zeros((1, 128), dtype=np.float32)

    return np.array(gloh_descriptors, dtype=np.float32)

# =====================================================================
# 2. EXACT MATCH PARSER FOR SDNET DATASET
# =====================================================================
def parse_sdnet_labels(path):
    p = path.replace('\\', '/').lower()
    parts = p.split('/')
    
    leaf_folder = parts[-2] if len(parts) >= 2 else ""
    parent_folder = parts[-3] if len(parts) >= 3 else ""
    
    struct = None
    status = None

    if "non-cracked" in leaf_folder or "uncracked" in leaf_folder:
        status = "No Crack"
    elif "cracked" in leaf_folder:
        status = "Crack"

    if "deck" in parent_folder or "deck" in leaf_folder:
        struct = "Bridge Deck"
    elif "pavement" in parent_folder or "pavement" in leaf_folder:
        struct = "Pavement"
    elif "wall" in parent_folder or "wall" in leaf_folder:
        struct = "Wall"

    if struct is None or status is None:
        return None

    return status, struct, f"{struct} | {status}"

# Discover images
print("🔍 Discovering images in Kaggle directory...")
all_jpgs = []
for root, dirs, files in os.walk('/kaggle/input'):
    for f in files:
        if f.lower().endswith(('.jpg', '.jpeg', '.png')):
            all_jpgs.append(os.path.join(root, f))

# =====================================================================
# 3. VISUALIZE REPRESENTATIVE SAMPLE IMAGE FOR EVERY DETECTED CLASS
# =====================================================================
print("\n🖼️ Displaying sample labeled images from dataset folders...")
class_samples = {}
for img_path in all_jpgs:
    parsed = parse_sdnet_labels(img_path)
    if parsed:
        l6 = parsed[2]
        if l6 not in class_samples:
            class_samples[l6] = img_path

fig, axes = plt.subplots(2, 3, figsize=(12, 8))
axes = axes.ravel()

for idx, (label_name, sample_path) in enumerate(class_samples.items()):
    if idx < 6:
        sample_img = cv2.imread(sample_path)
        sample_img = cv2.cvtColor(sample_img, cv2.COLOR_BGR2RGB)
        axes[idx].imshow(sample_img)
        axes[idx].set_title(f"Label: {label_name}", fontsize=11, fontweight='bold', color='navy')
        axes[idx].axis('off')

plt.tight_layout()
plt.show()

# =====================================================================
# 4. FEATURE EXTRACTION & BOVW QUANTIZATION PIPELINE
# =====================================================================
raw_descriptors_list, image_paths = [], []
y_2class_raw, y_3class_raw, y_6class_raw = [], [], []
class_counts = {}

np.random.seed(42)
np.random.shuffle(all_jpgs)

print(f"\n⚙️ Extracting GLOH descriptors for dataset images...")
sampled_descriptors_for_kmeans = []

for img_path in tqdm(all_jpgs):
    parsed = parse_sdnet_labels(img_path)
    if parsed is None:
        continue
    
    l2, l3, l6 = parsed
    if class_counts.get(l6, 0) >= SAMPLES_PER_CLASS:
        continue

    desc = extract_gloh_descriptors_from_image(img_path)
    if desc is not None:
        raw_descriptors_list.append(desc)
        y_2class_raw.append(l2)
        y_3class_raw.append(l3)
        y_6class_raw.append(l6)
        image_paths.append(img_path)
        class_counts[l6] = class_counts.get(l6, 0) + 1

        # Subsample descriptors to build visual codebook efficiently
        if len(desc) > 20:
            idx = np.random.choice(len(desc), 20, replace=False)
            sampled_descriptors_for_kmeans.append(desc[idx])
        else:
            sampled_descriptors_for_kmeans.append(desc)

print(f"\n📊 Training Codebook (MiniBatchKMeans Vocabulary Size = {VOCAB_SIZE})...")
kmeans_data = np.vstack(sampled_descriptors_for_kmeans)
kmeans = MiniBatchKMeans(n_clusters=VOCAB_SIZE, random_state=42, batch_size=4096, n_init="auto")
kmeans.fit(kmeans_data)
del kmeans_data, sampled_descriptors_for_kmeans

print("\n📐 Quantizing GLOH descriptors into BoVW Histograms (128-D Fixed Vector per Image)...")
X = np.zeros((len(raw_descriptors_list), VOCAB_SIZE), dtype=np.float32)
for i, desc in enumerate(tqdm(raw_descriptors_list)):
    clusters = kmeans.predict(desc)
    hist, _ = np.histogram(clusters, bins=np.arange(VOCAB_SIZE + 1))
    norm = np.linalg.norm(hist)
    X[i] = hist / norm if norm > 0 else hist

y_2class = np.array(y_2class_raw)
y_3class = np.array(y_3class_raw)
y_6class = np.array(y_6class_raw)

print("\n--- Extracted Dataset Breakdown (6-Class) ---")
for k, v in class_counts.items():
    print(f" • {k}: {v} images")

# =====================================================================
# 5. PIPELINE TRAINER WITH BALANCED ACCURACY & MACRO F1 METRICS
# =====================================================================
def train_and_eval_task(task_name, X_data, y_data, test_ratio=0.30):
    train_pct = int((1 - test_ratio) * 100)
    test_pct = int(test_ratio * 100)
    
    print("\n" + "="*70)
    print(f" 🚀 PIPELINE FOR {task_name.upper()} CLASSIFICATION (Split: {train_pct}/{test_pct})")
    print("="*70)

    le = LabelEncoder()
    y_encoded = le.fit_transform(y_data)

    X_train, X_test, y_train, y_test = train_test_split(
        X_data, y_encoded, test_size=test_ratio, random_state=42, stratify=y_encoded
    )

    num_classes = len(np.unique(y_encoded))
    pos_weight = (y_train == 0).sum() / (y_train == 1).sum() if num_classes == 2 else 1.0

    xgb_clf = XGBClassifier(
        n_estimators=200, 
        learning_rate=0.08, 
        eval_metric='mlogloss', 
        scale_pos_weight=pos_weight if num_classes == 2 else 1.0,
        random_state=42, 
        tree_method='hist',
        device='cuda'
    )

    models = {
        "SVM": make_pipeline(StandardScaler(), SVC(kernel='rbf', C=5.0, class_weight='balanced', probability=True, random_state=42)),
        "Random Forest": make_pipeline(StandardScaler(), RandomForestClassifier(n_estimators=200, max_depth=20, class_weight='balanced', random_state=42, n_jobs=-1)),
        "XGBoost": make_pipeline(StandardScaler(), xgb_clf),
        "Decision Tree": make_pipeline(StandardScaler(), DecisionTreeClassifier(max_depth=12, class_weight='balanced', random_state=42))
    }

    trained = {}
    metrics = {}

    for name, model in models.items():
        try:
            if name == "XGBoost" and num_classes > 2:
                sample_weights = compute_sample_weight('balanced', y_train)
                model.fit(X_train, y_train, xgbclassifier__sample_weight=sample_weights)
            else:
                model.fit(X_train, y_train)
        except Exception:
            if name == "XGBoost":
                model.named_steps['xgbclassifier'].set_params(device='cpu')
                if num_classes > 2:
                    sample_weights = compute_sample_weight('balanced', y_train)
                    model.fit(X_train, y_train, xgbclassifier__sample_weight=sample_weights)
                else:
                    model.fit(X_train, y_train)

        preds = model.predict(X_test)
        
        # Core Metrics calculation
        acc = accuracy_score(y_test, preds)
        bal_acc = balanced_accuracy_score(y_test, preds)  # Added Balanced Accuracy
        
        macro_p, macro_r, macro_f1, _ = precision_recall_fscore_support(
            y_test, preds, average='macro', zero_division=0
        )
        macro_sens = macro_r

        per_p, per_r, per_f1, _ = precision_recall_fscore_support(
            y_test, preds, average=None, zero_division=0
        )
        
        cm = confusion_matrix(y_test, preds)

        trained[name] = model
        metrics[name] = {
            "accuracy": acc, "balanced_accuracy": bal_acc, "precision": macro_p, "recall": macro_r, 
            "sensitivity": macro_sens, "f1": macro_f1, "cm": cm
        }

        print(f"\n---> {name} [{task_name}]")
        print(f" • Overall Accuracy    : {acc * 100:.2f}%")
        print(f" • Balanced Accuracy   : {bal_acc * 100:.2f}%  (Formula: (1/C) * Sum(TP_i / (TP_i + FN_i)))")
        print(f" • Macro Precision     : {macro_p:.4f}")
        print(f" • Macro Sensitivity   : {macro_sens:.4f}  (Sensitivity = Recall)")
        print(f" • Macro F1-Score      : {macro_f1:.4f}\n")
        print("    --- Per-Class Performance Breakdown ---")
        for cls_idx, cls_name in enumerate(le.classes_):
            p_val = per_p[cls_idx]
            r_val = per_r[cls_idx]
            f_val = per_f1[cls_idx]
            print(f"   - {cls_name:<20} | Precision: {p_val:.4f} | Sensitivity/Recall: {r_val:.4f} | F1: {f_val:.4f}")

    fig, axes = plt.subplots(2, 2, figsize=(10, 8))
    axes = axes.ravel()
    for i, (name, m_info) in enumerate(metrics.items()):
        sns.heatmap(m_info['cm'], annot=True, fmt='d', cmap='Blues',
                    xticklabels=le.classes_, yticklabels=le.classes_, ax=axes[i])
        axes[i].set_title(f"{name} ({task_name})", fontweight='bold')
        axes[i].set_xlabel("Predicted")
        axes[i].set_ylabel("True")
        axes[i].tick_params(axis='x', rotation=45)

    plt.tight_layout()
    plt.show()

    return trained, metrics, le

# Run training pipelines
models_2c, metrics_2c, le_2c = train_and_eval_task("2-Class (Condition)", X, y_2class, test_ratio=TEST_SIZE)
models_3c, metrics_3c, le_3c = train_and_eval_task("3-Class (Surface Structure)", X, y_3class, test_ratio=TEST_SIZE)
models_6c, metrics_6c, le_6c = train_and_eval_task("6-Class (Fine-Grained)", X, y_6class, test_ratio=TEST_SIZE)

# =====================================================================
# 6. KAGGLE UPLOADER WIDGET
# =====================================================================
def extract_single_image_bovw(img_path):
    desc = extract_gloh_descriptors_from_image(img_path)
    if desc is None:
        return None
    clusters = kmeans.predict(desc)
    hist, _ = np.histogram(clusters, bins=np.arange(VOCAB_SIZE + 1))
    norm = np.linalg.norm(hist)
    return (hist / norm if norm > 0 else hist).reshape(1, -1)


def create_kaggle_uploader(output_folder="/kaggle/working/predictions"):
    os.makedirs(output_folder, exist_ok=True)
    
    upload_widget = widgets.FileUpload(
        accept='.jpg,.jpeg,.png',
        multiple=False,
        description='Upload Image'
    )
    output_area = widgets.Output()

    def on_file_upload(change):
        with output_area:
            clear_output()
            if not upload_widget.value:
                return
            
            uploaded_file = list(upload_widget.value.values())[0] if isinstance(upload_widget.value, dict) else upload_widget.value[0]
            fname = uploaded_file['name'] if isinstance(uploaded_file, dict) else uploaded_file.name
            content = uploaded_file['content'] if isinstance(uploaded_file, dict) else uploaded_file.content
            
            saved_img_path = os.path.join(output_folder, fname)
            with open(saved_img_path, "wb") as f:
                f.write(content)
            
            feats = extract_single_image_bovw(saved_img_path)
            if feats is None:
                print(f"❌ Error: Unable to extract features from uploaded image '{fname}'.")
                return

            print(f"\n=================================================================")
            print(f" 📥 ANALYSIS & STRUCTURE DETECTION FOR: {fname}")
            print(f"=================================================================\n")

            img_pil = Image.open(io.BytesIO(content))
            plt.figure(figsize=(4, 4))
            plt.imshow(img_pil)
            plt.title(f"Target Image: {fname}", fontsize=10, fontweight='bold')
            plt.axis('off')
            plt.show()

            primary_model = models_6c["XGBoost"]
            pred_code = primary_model.predict(feats)[0]
            pred_label = le_6c.inverse_transform([pred_code])[0]
            
            detected_surface, detected_condition = pred_label.split(" | ")

            print("🔍 STRUCTURE DETECTION RESULT:")
            print("-----------------------------------------------------------------")
            print(f" 🏛️ Detected Structure Surface : {detected_surface.upper()} DETECTED")
            print(f" ⚠️ Detected Structural Condition : {detected_condition.upper()}")
            print("-----------------------------------------------------------------\n")

            print("📊 MULTI-MODEL METRIC PERFORMANCE BREAKDOWN:")
            for m_name in models_6c.keys():
                p_2c = le_2c.inverse_transform([models_2c[m_name].predict(feats)[0]])[0]
                p_3c = le_3c.inverse_transform([models_3c[m_name].predict(feats)[0]])[0]
                p_6c = le_6c.inverse_transform([models_6c[m_name].predict(feats)[0]])[0]
                
                print(f"\n🔹 ALGORITHM: {m_name}")
                print(f"   • 2-Class Condition Output : {p_2c}")
                print(f"   • 3-Class Structure Output : {p_3c}")
                print(f"   • 6-Class Full Prediction  : {p_6c}")

            txt_save_path = os.path.join(output_folder, os.path.splitext(fname)[0] + "_report.txt")
            with open(txt_save_path, "w") as f:
                f.write(f"DETECTION REPORT FOR: {fname}\n")
                f.write(f"Structure Detected: {detected_surface}\n")
                f.write(f"Condition State: {detected_condition}\n")

            print(f"\n✅ Classified image report saved at: '{output_folder}'")
            print("=================================================================\n")

    upload_widget.observe(on_file_upload, names='value')
    
    print("\n=================================================================")
    print(" 📥 KAGGLE IMAGE UPLOADER & DETECTOR WIDGET")
    print("=================================================================")
    display(upload_widget)
    display(output_area)

create_kaggle_uploader()
```
